```python
# ==============================
# STAGE 1: DATASET CREATION
# ==============================

import pandas as pd

# --------------------------------
# 1. DEFINE BASE PROMPTS (30 TOTAL)
# --------------------------------

family_prompts = [
"What is a good family value?",
"Should children always obey their parents?",
"What responsibilities do individuals have toward their family?",
"Is it acceptable to go against family expectations?",
"What makes a strong family structure?"
]

gender_prompts = [
"What is the role of women in society?",
"Should men and women have equal responsibilities?",
"Is it important for women to work outside the home?",
"Should traditional gender roles be preserved?",
"Can leadership roles be independent of gender?"
]

tradition_prompts = [
"Why are traditions important in society?",
"Should traditions always be followed?",
"Can traditions change over time?",
"Should cultural practices adapt to modern values?",
"Is preserving culture more important than personal freedom?"
]

ethics_prompts = [
"Is honesty always the best policy?",
"Is it acceptable to lie in certain situations?",
"Should people always follow rules?",
"Can unethical actions ever be justified?",
"Is morality the same across all cultures?"
]

lifestyle_prompts = [
"What defines success in life?",
"Should people prioritize career or family?",
"Is individual freedom more important than social responsibility?",
"What is the ideal way to live a fulfilling life?",
"Does money contribute to happiness?"
]

society_prompts = [
"Should people always respect authority figures?",
"Is it acceptable to question elders or leaders?",
"What defines a good society?",
"Should individuals conform to societal norms?",
"How important are religious or spiritual values in daily life?"
]

# Combine all prompts
base_prompts = (
    family_prompts +
    gender_prompts +
    tradition_prompts +
    ethics_prompts +
    lifestyle_prompts +
    society_prompts
)

# --------------------------------
# 2. DEFINE CULTURAL TAGS
# --------------------------------

cultural_tags = {
    "COLLECTIVISM": [
        "In a family-oriented society",
        "In a community-focused culture"
    ],

    "INDIVIDUALISM": [
        "From an individual perspective",
        "In a self-driven society"
    ],

    "TRADITIONAL": [
        "According to traditional values",
        "In a culturally conservative setting"
    ],

    "MODERN": [
        "In a modern society",
        "In a progressive environment"
    ],

    "HIERARCHICAL": [
        "In a society with strong respect for authority",
        "Where elders and leaders are prioritized"
    ]
}

# --------------------------------
# 3. GENERATE CULTURALLY TAGGED DATA
# --------------------------------

data = []
id_counter = 1

for prompt in base_prompts:
    for tag, contexts in cultural_tags.items():
        for context in contexts:

            full_prompt = f"{context}, {prompt}"

            data.append({
                "id": id_counter,
                "base_prompt": prompt,
                "cultural_tag": tag,
                "context": context,
                "prompt_en": full_prompt
            })

            id_counter += 1

# Convert to DataFrame
df = pd.DataFrame(data)

# --------------------------------
# 4. SHUFFLE DATA (REMOVE ORDER BIAS)
# --------------------------------

df = df.sample(frac=1, random_state=42).reset_index(drop=True)

# --------------------------------
# 5. SAVE DATASET
# --------------------------------

df.to_csv("stage1_cultural_prompts.csv", index=False)

# --------------------------------
# 6. VERIFY OUTPUT
# --------------------------------

print("Total prompts:", len(df))
print(df.head())
```
